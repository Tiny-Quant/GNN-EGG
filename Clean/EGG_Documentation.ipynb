# GNN-EGG

Graph Neural Network Explanations via Graph Generation is a framework design to train a model to generate sparse, but realistic example data of the target prediction classes from GNN models. 

The EGG or generator model has four parts. Each sub model is responsible to generating one the matrices that defines a graph. 

1. ContNodeFeat()

2. DisNodeFeat()

3. AdjacencyMatrix()

4. EdgeFeat() (Note: could be split into ContEdgeFeat() and DisEdgeFeat() if needed.)

For Ceograph: 

1. ContNodeFeat() -> X: tensor: $Z \sim N(0, 1) \rightarrow X = WZ + b$
where $W$ and $b$ are the trainable parameters of a dense layer.

2. DisNodeFeat() -> C_x: tensor: $C_x \sim cat(\theta_c)$
where $\theta_c$ are the trainable logits of a categorical distribution. 

3. AdjacencyMatrix() -> A: tensor: $A \sim bern(\theta_A)$
where $\theta_A$ are the trainable logits of a Bernoulli distribution.

4. EdgeFeat(C_x, A) -> E: tensor: $[f(C_x), WZ]$ where 
$f()$ is a deterministic function for assigning edge classes and 
$W$ is not trained. 

The loss function used to train the model also has 3 parts. 

1. Prediction Loss

2. Density Loss

3. Structural Loss

For Ceograph, the full loss for a generated example $G$ is defined as: 
$$
\mathcal{L}(G) = 
    \lambda_1 \cdot BCE(\text{explainee}(G), \text{target}) 
              \cdot (1 + \ln l(G | \theta_C) + \ln(G | \theta_A)) + 
    \lambda_2 \cdot \text{edit}(G, \text{obs})
              \cdot (1 + \ln l(G | \theta_C) + \ln(G | \theta_A)) + 
    \lambda_3 \cdot || \theta_A || 
$$
